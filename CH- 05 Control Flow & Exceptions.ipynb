# 1. Recursion

Basic way to define:- 

- Function calling itself again and again 

Better way to define:- 
- When a problem can be broken down into smaller problems and ultimately reaching an end case, we use recursion. 

```python
def fact(n):
    if n == 1: return 1
    return n * fact(n-1)

fact(4)
```

##  Internal Working of recursion. 
Easy way to understand. 
1.  We go down down, fact(n-2), fact(n-3), bascially break into different smaller problems. 
2.  we come up & up, and compute the values. 

Downward process:- It uses the Call Stack, which grows and grows

Upward process:- Stack Starts Unwinding, `computation begins` 1 * `2`, 2 * `3`, 6 * `4`

## Is recurssion faster?
* Not exactly, infact it is an expensive operation, uses lots of memory (due to `call stacks` )

## But why then? 
* Easy to visualize the problem breaking to smaller ones. 
* When used with `memoization` it works really faster, (you are not computing same thing again)
* bascially in dynamic programming it helps. 


## Recursion vs Iteration
* Iteration of `loops` are faster and use lesser memory, (not call stack so.. )

Q. Why does Python raise RecursionError?

Because each recursive call consumes stack space, and CPython enforces a maximum recursion depth to prevent stack overflow.

Q. When would you prefer recursion over iteration?

When the problem is naturally recursive, such as tree traversal, graph DFS, nested structures, or backtracking.

# 2 Context Managers

```python
with open("data.txt") as f:
    data = f.read()
```

### why do we use `with` and open files this way???

```python
# consider this case: 
f = open("data.txt")
data = f.read()

x = 1/0 ## should raise error 
f.close() # ! ???? 
```

* Two things are going to happen here: 
1. Divsion by zero error. 
2. And so the rest of code stops. 

- And so, `file never closes`. 
- This is called `resource leak`. 

This will cost alot of issues with
* Files
* Database connections
* Network sockets
* GPU Memory
* Threads


Whenever we acquire any resource, we should release it, even if any error occurs. 

This is why `With` is used to handle these issues. 

```python
with open("data.csv") as f:
    data = f.read()
```

- When this block finishes, python automatically `close the file`, even if exception occurs. 

- Entire purpose of `with` is to `acquire resource`, `use it`, and `gurantee its cleanup`. 

## Internal working of `with`

```python
with open("data.txt") as f:
    data = f.read()

## This is conceptully similar to writing as follows:- 

obj = open("data.txt")

obj.__enter__()
try: 
    data = obj.read()
finally: 
    obj.__exit__()
```

### What is `__enter__()` and `__exit__()`??

* These are context manager protocol. 

so when we write `with open(path) as f: `, python internally does
`f = file.__enter__()`. 

and `file.__exit__()` to clean up the job, irrespctive of any `error encountered`.
