### What is a decorator?

* so it is a function, taking another function as input, making some additional changes without altering the `source code` of main function .


```python
# Eg:- 

def make_masala_dosa(func): # Modifies functionality: simple dosa --> masala dosa. 
    pass

@make_masala_dosa
def make_dosa(): #
    pass
```

## Why use decorator?
### why not just modify the existing function only?

* So functions are meant for single responsibility task.
* To reduce the repeatability

```python
## eg: let's write a function to make tea. 

def make_tea():
    print("Started...")
    import time
    time.sleep(1)
    print("Tea Ready")
make_tea()
```

Now, If we want to measure the time taken in preparing the tea, 
we can write the code in itself only, but `time measuring` can be resued in future with some more functions right. 
So, better to separately write this functionality. 

```python
import time
```

```python
def timer(base_func):

    def enhanced_fn():
        start_time = time.time()
        base_func()
        end_time = time.time()
        print(f"total time taken: {end_time - start_time}")
    return enhanced_fn

def make_tea():
    print("Started...")

    time.sleep(1)
    print("Tea Ready")

time_to_tea = timer(make_tea)
time_to_tea()
```

```python
### Or using the decorator syntax
```

```python
def timer_fn(base_func):

    def enhanced_fn():
        start_time = time.time()
        base_func()
        end_time = time.time()
        print(f"total time taken: {end_time - start_time}")
    return enhanced_fn

@timer_fn # ! decorator
def make_tea():
    print("Started...")

    time.sleep(1)
    print("Tea Ready")

make_tea()
```

```python
### Same way we can add this decorator to any other function to measure the time of execution. 
```

```python
@timer_fn # measures the timing. 
def multiply_numbers():
    x = 4
    y = 325.321
    print(x*y + x**2 + y**3 + (x*y)**2)
    time.sleep(1)
multiply_numbers()
```

## What if one function has some arguments. 

```python
@timer_fn
def add_values(x, y):
    sum = x+ y
    print(sum)

add_values(3,5)
```

* ## Error...

* Need to make the decorator `flexible` enough to handle any arguement issues
* So We need to use the `*args` and `**kwargs`. 
* this works even if we need/don't need the parameters for our base function. 

```python
def timer_fn(base_func):

    def enhanced_fn(*args, **kwargs):
        start_time = time.time()
        base_func(*args, **kwargs) # ! *args & **Kwargs 
        end_time = time.time()
        print(f"total time taken: {end_time - start_time}")
    return enhanced_fn

@timer_fn
def add_values(x, y):
    sum = x+ y
    time.sleep(1)
    print(sum)

add_values(3,5)

```

### Qns:- 
1. Decorator modifies the original function?
- - They create a new `wrapper function` and return it. 

2. Decorator only works because of `@`?
- - No, things can be written without it, `dec_fn = wrapper(base_function)`

* Decorator help implement `DRY principle` by separating functionalities like (`logging`, `caching`, `timing`, `authentication`, ) from business logic.
