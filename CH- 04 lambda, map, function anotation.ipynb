## List sorting

```python
## to sort any dictionary/Mulit-dimension list basis, either `key` or `value`

students = [
    ("Zlice", 90),
    ("Loob", 70),
    ("Marlie", 85)
]

print(sorted(students, key =  lambda x: x[0])) # as per name, ascending
print(sorted(students, key = lambda x: -1*x[1])) # as per values, descending
```

## Dictionary Sorting

```python
scores = {
    "Alice": 90,
    "Bob": 70,
    "Charlie": 85
}

print(sorted(scores.items(), key = lambda x: x[1]))
# sorted(scores, key = lambda x: x[1]) # This takes `keys` only. 
```

```python
## IT sorts only the keys, if .items not mentioned. 
scores = {
    "Alice": 90,
    "Bob": 70,
    "Charlie": 85
}

print(sorted(scores, key = lambda x: x[1]))
```

# Map, Reduce, Filter

`map()` -> transform every element  
`filter()` -> keep some element  
`reduce()`  -> get single value

```python
# map(function, iterable)
x = [2, 4, 6]
x = map(lambda x: int(x/2), x)
print(list(x))
```

```python
## Filter:
# Filter(function, iterable)
# ! Filter strings or integers. 
x = ['a', 'b', 2,4,6, '4']
y = filter(lambda x: isinstance(x, str), x)
print(list(y))
```

```python
## reduce. 
from functools import reduce
# reduce(function, iterable)
num = [1,23,4,6,42,2,1]
reduce(lambda x, y: x+y , num)
```

## Why do `map()` and `filter()` return iterators?

* BTW, what is an `iterator`?
- object, that fetches eleemnts `one by one`. 
* `Iterable`?
- a container or sequence you can loop over. `list, tuple, dict..`

```python
map(lambda x: x**2, num)
## v/s list(map(.....))
```

* Python doesn't create entire result immediately. 
* It generate iterator that generate values on demand. `lazy evaluation`

* why Use iterators?
-` Memory efficient.` compute and fetch data on demand, hold one item in memory at a time. good for massive data file. 
- 

`list comprehension` are more Pythonic and easy to understand than using `map() or filter()`.

```python
nums =  [2,24,5,6,7,8,3]

# Using list comprehension. 
x = [x  for x in nums if x %2]
print(x)

# using filter
x = filter(lambda x: x%2, nums)
print(list(x))
```

### Pydantic Example

* but what is `pydantic`??
- so its a `data validation` and `parsing` `library`, that enforces type hints at runtime. 
- So it ensure that data structures like `api input`, `database results`, are valid and correctly typed, before your application process them. 

```python
## eg: - 
from pydantic import BaseModel, EmailStr

class User(BaseModel):
    id: int
    name: str
    email: str
    is_active: bool = True, # default value, optional

external_data = {
    "id": "2134", ## Will be coerced from string -> integer. 
    "name": 433433434 ,  #! Will throw error. 
    "email": 222 #! Will throw error. 
}

user = User(**external_data)
print(user.id)
```

1. ValidationError: 2 validation errors for User
2. Input should be a valid string
