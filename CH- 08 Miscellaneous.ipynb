# Iterator & Generator

* iterator - this object let's access items in a sequence one at a time. 
* doesnot load entire data at once, this saves memory and follows lazy evaluation. 

````python
listt = iter(["ramu", '3443', 434, True])
print(next(listt))
print(next(listt))
print(next(listt))
print(next(listt))
````

````python
# so when we write a for loop. 
data = [43,3,2,4,4,0,32]
for x in data:
    print(x)
````

````python
 # !python is doing somethign like> 

iterator = iter(data)
while True:
    try: 
        x = next(iterator)
        print(x)
    except StopIteration:
        break
````

Generator is another way to create iterator, in simpler word using `yield`

### All generator are iterator
- But not all iterator are generator. 

As you can build custom iterator using class with `__iter__()` and `__next__()` methods, `without using any` yield / generator. 

* So `next` remember its position and doesnot restart . 

* Generator: allow us to produce sequence of vlaues on demand(`lazy) `instead of storing them all in the memory at once. 

* use `yield` keyword. 

````python
def count_(mx_val):
    count = 1
    while count <= mx_val:
        yield count
        count += 1

cnt = count_(5)
print(next(cnt))
````

````python
print(next(cnt))
print(next(cnt))
print(next(cnt))
print(next(cnt))
````

````python
print(next(cnt))
````

### why generator matters

````python
# suppose you write:- 
def get_data():
    return [huge_dataset]

````

* Here everything is held in the memory

````python
# INstead we can write this as:- 
def get_data():
    for row in huge_dataset:
        yield row
````

* So `lazy evaluation` and `memory efficiency`, especially for large or streaming data.

* Used in   `infinite stream`. 

````python
def infinite_sequence():
    num = 0
    while True:
        yield num
        num += 1
````

generate continous numbers safely because values are calculated only when requested.

# Concurrency, GIL & Async

* why do we need to learn about them??? 

- when some part of our code is doing job, others have to wait for it finish...
- so when our website is loading/ downloding file,,, we'll stuck there,.. can't do other things because of it.
-

So there comes the concept of `thread`. 
- thread is like a worker doing a job in kitchen . 
- he does one task at a time, if he is boiling water, others will have to wait for doing other tasks. 

So we've `Input/Output` bound tasks and `cpu-bound` tasks 
- Input/ output is where some tasks run when one is waiting. 

#### I/O-Bound Tasks: 

- Leave the processor idle during waits.
- Scale well with asynchronous code or multiple threads.
- Examples include `downloading files,` querying a database, or `reading a hard drive.`


#### CPU-Bound Tasks
- that which `require lots of cpu time.`.. 
- cpu based require heavy computing, video rendering/editing..
- Keep the processor busy 100 percent of the time.
- Scale best with multiple processor cores or parallel processing.- Examples include video rendering, image processing, and heavy data crunching.

GIL: `Global Interpreter lock.` 
- It ensures only 1 thread executes Python bytecode at any moment. 
- one python interpretor and memory in work. 

- even if the computer has 16 cpu cores, a standard threaded python program running heavy calculatoin will use only 1 core because of GIL. 

### concurrency v/s parallelism. 


- concurrency is dealing with many things: Task A -> [Pause] -> Task B -> [Pause] -> Task A


parallelism (doing many things at once): 
```
Task A (core 1), 
Task B (core 2)
```

* Parallelism:- multiple tasks at `exact same physical milisecond `

### How to achieve concurrency in Python. .

#### 1. Asyncio: 
- when one task hits waiting point (eg. downloading a file), it voluntarily yields controls back, which moves to the next task.  . 
- best for `web scraping` & `API calls`. 

````python
import asyncio
import time 

# `async def` to define a non-blocking asynchronous function. 

async def fetch_api_data(endpoint_id: int) -> dict:
    print(f"[Async] Starting request for ID: {endpoint_id}")

    # ! await tells event loop, i'm waiting, go run another task 
    await asyncio.sleep(1.5) 

    print(f"finished request for ID {endpoint_id}")
    return {"id": endpoint_id, "status": "sucess"}

async def main():
    start_time = time.perf_counter()
    tasks = [fetch_api_data(i) for i in range(1, 4)]

    results = await asyncio.gather(*tasks)

    end_time = time.perf_counter()
    print(f"All API task completed in {end_time - start_time} seconds")

    print(f"Results: {results}")

if __name__ == "__main__":
    asyncio.run(main())
````

#### 2. Threading
* best for background I/O Operations
* whenever one is waiting, other will execute tasks. 

````python
from concurrent.futures import ThreadPoolExecutor
import time

def save_to_database(record_id: int) -> str:
    print(f"[Thread] Writing record {record_id} to database...")
    time.sleep(4)  #! Simulates network/disk delay
    print(f"[Thread] Record {record_id} saved.")
    return f"Record {record_id} OK"

def main():
    start_time = time.perf_counter()
    records = [101, 102, 103]
    
    #!  Context manager automatically handles closing and cleaning up threads
    with ThreadPoolExecutor(max_workers=3) as executor:
        # 'map' sends the items in the list to the function across different threads
        results = list(executor.map(save_to_database, records))
        
    end_time = time.perf_counter()
    print(f"All database threads completed in {end_time - start_time:.2f} seconds.")
    print(f"Results: {results}")

if __name__ == "__main__":
    main()
````

* So, we don't have to wait, when the disk is dealayed `(time.sleep)`, 
* executor is running on `3` different threads `(Writing record 101/102/103 to database...)` ran all together. 

#### 3. Parallelism

- Each process has its own memory space, 

````python
from concurrent.futures import ProcessPoolExecutor
import time

def heavy_math_calculation(number: int) -> int:
    print(f"[Process] Calculating factorial/matrix for {number}...")
    # Simulate a heavy CPU calculations loop
    total = 0
    for i in range(10_000_000):
        total += i
    print(f"[Process] Finished calculation for {number}.")
    return total + number

def main():
    start_time = time.perf_counter()
    numbers_to_process = [50, 60, 70, 80]
    
    # ! Spawns separate system processes, each with its own Python interpreter and GIL
    with ProcessPoolExecutor() as executor:
        results = list(executor.map(heavy_math_calculation, numbers_to_process))
        
    end_time = time.perf_counter()
    print(f"All CPU calculations finished in {end_time - start_time:.2f} seconds.")
    print(f"Results: {results}")

# !IMPORTANT: Multiprocessing requires this block protection in Python
if __name__ == "__main__":
    main()
````
